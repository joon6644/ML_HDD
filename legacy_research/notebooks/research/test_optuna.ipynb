# Optuna 단계별 트라이얼 테스트

`src.train_core.run_training`을 호출하여 Stage 1과 Stage 2의 트라이얼 횟수를 각각 지정하여 테스트하는 노트북입니다.

```python
import sys
import os
from pathlib import Path

notebook_dir = Path.cwd()
project_root = notebook_dir.parent.parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from config import train_config as cfg
from src.train_core import run_training, plot_confusion_matrix, print_ensemble_summary, plot_subset_prauc
```

## 1. 모델 학습 및 Optuna 최적화 (Stage별 횟수 지정)

```python
# Stage 1과 Stage 2의 횟수를 각각 2회, 1회로 지정하여 총 3회를 테스트합니다.

result_dict = run_training(
    cfg=cfg,
    feature_cols=cfg.FEATURE_COLS,
    run_optuna=True,
    optuna_s1_trials=2,             # Stage 1: 2회
    optuna_s2_trials=1,             # Stage 2: 1회
    interactive_rerank=True,        # 리랭크 전 표 출력 및 대기
    cleanup_optuna_temp=False       # 실험을 위해 모델 보존
)
```

## 2. 결과 점검 및 시각화

```python
ens_result = result_dict["ensemble_result"]
best_params = result_dict["best_params"]
print_ensemble_summary(ens_result)
plot_subset_prauc(ens_result)
plot_confusion_matrix(ens_result, cfg)
```
