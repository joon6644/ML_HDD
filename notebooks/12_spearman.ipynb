```python
import warnings
import logging

# 1. 모든 경고 메시지 무시 (FutureWarning, UserWarning 등)
warnings.filterwarnings('ignore')

# 2. matplotlib 및 기타 라이브러리의 시스템 로그 레벨 조정
logging.getLogger('matplotlib').setLevel(logging.ERROR)

import duckdb
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. 데이터 로드
path = r'C:\Workspace\06_ML_projdect\26_1_COIN\data\rfe_sample_data\rfe_sample_train.parquet'
con = duckdb.connect()

print(f"데이터 로드 중: {path}")
df = con.execute(f"SELECT * FROM read_parquet('{path}')").df()

# 2. 피처 그룹 정의
feature_groups = {
    "최근 부하 압력": [
        "is_warmup_14d", "is_warmup_7d", "workload_intensity", 
        "age_weighted_workload", "workload_7d_accel"
    ],
    "총 탐색량": [
        "total_seeks_28d_asfd", "total_seeks_28d_max", 
        "total_seeks_14d_mean", "total_seeks_7d_std"
    ],
    "총 읽기량": [
        "s242_28d_asfd", "s242_28d_std", "s242_28d_sum", "s242_28d_max", 
        "s242_diff", "smart_242_raw"
    ],
    "총 기록량": [
        "s241_28d_asfd", "s241_28d_std", "s241_28d_mean", "s241_28d_max", 
        "s241_diff", "smart_241_raw"
    ],
    "누적 사용량": [
        "smart_9_raw"
    ],
    "Burst 이상": [
        "s5_relative_score_14d", "s192_14d_burst", "s189_28d_highfly_burst", 
        "s187_14d_burst_index", "read_spike_ratio", "s189_28d_sum", 
        "s189_28d_max", "s187_14d_sum", "s187_14d_max", 
        "s189_diff", "smart_189_raw"
    ],
    "Reallocated / Pending": [
        "s5_relative_score_14d", "reallocated_pending_ratio", "pending_to_offline_ratio"
    ],
    "급성 Spike": [
        "read_spike_ratio", "write_spike_ratio", "uncorrectable_spike_ratio", 
        "seek_spike_ratio", "seek_error_14d_spike_ratio"
    ],
    "Sector 열화": [
        "s187_14d_sum", "s187_14d_max", "s198_error_rate", "s187_error_rate", 
        "error_density_14d", "s197_recovery_flag", "multi_error_coincidence", 
        "error_saturation_score", "cumulative_error_score", "s198_28d_sum", 
        "s198_28d_max", "s198_14d_sum", "s198_14d_max", "s197_28d_sum", 
        "s197_28d_max", "s197_14d_sum", "s197_14d_max", "s187_28d_sum", 
        "s187_28d_max", "smart_187_raw", "s187_diff", "smart_197_raw", 
        "s197_diff", "smart_198_raw", "s198_diff"
    ],
    "읽기/쓰기 안정성": [
        "write_stability_ratio", "s183_28d_sum", "s183_28d_max", 
        "s183_14d_sum", "s183_14d_max", "s183_diff", "smart_183_raw"
    ],
    "Seek 경로 이상": [
        "age_weighted_seek_error", "timeout_seek_density", "seek_error_density", 
        "seek_error_28d_sum", "seek_error_28d_max", "seek_error_14d_sum", 
        "seek_error_14d_max", "seek_error_count_raw", "seek_error_count_diff"
    ],
    "기본 I/O 이상": [
        "s199_days_since_last", "timeout_to_uncorrectable_lag1", "timeout_read_density", 
        "s199_error_density", "io_asymmetry_index", "s199_14d_burst", 
        "timeout_total_28d_sum", "timeout_total_28d_max", "timeout_total_14d_sum", 
        "timeout_total_14d_max", "timeout_5s_28d_sum", "timeout_5s_28d_max", 
        "timeout_5s_14d_sum", "timeout_5s_14d_max", "s199_28d_sum", 
        "s199_28d_max", "s199_14d_sum", "s199_14d_max", "timeout_5s_raw", 
        "timeout_5s_diff", "timeout_total_raw", "timeout_total_diff", 
        "s199_diff", "smart_199_raw"
    ],
    "물리 스트레스 상호작용": [
        "log_shock_fly_interaction", "fatal_crash_interaction", "shock_to_highfly_ratio"
    ],
    "기계적 충격": [
        "s191_days_since_last", "shock_seek_interaction", "shock_fatigue_rate", 
        "s191_28d_sum", "s191_28d_max", "s191_14d_sum", "s191_14d_max", 
        "s191_diff", "smart_191_raw"
    ],
    "열 스트레스": [
        "temp_error_index", "thermal_fatigue_integral_7d", "s190_28d_mean", 
        "s190_diff", "smart_190_raw"
    ],
    "온도 수준": [
        'thermal_stress_index', 's194_over40_7d_count', 's194_28d_ewma', 's194_28d_dai', 's194_28d_cid', 's194_28d_asfd', 's194_28d_zscore', 's194_28d_std', 's194_28d_mean', 's194_28d_max', 's194_14d_ewma', 's194_14d_dai', 's194_14d_cid', 's194_14d_asfd', 's194_14d_zscore', 's194_14d_std', 's194_14d_mean', 's194_14d_max', 's194_7d_ewma', 's194_7d_dai', 's194_7d_cid', 's194_7d_asfd', 's194_7d_zscore', 's194_7d_std', 's194_7d_mean', 's194_7d_max', 's194_diff', 'smart_194_raw'
    ],
    "시스템성 실패": [
        "cascading_failure_flag", "zero_to_hero_count", "s184_1d_crash_flag", 
        "data_corruption_hazard", "s184_14d_sum", "s184_14d_max", 
        "s184_7d_sum", "s184_7d_max", "s184_3d_sum", "s184_3d_max", 
        "smart_184_raw", "s184_diff"
    ],
    "펌웨어 실패": [
        "firmware_struggle_index"
    ],
    "지속 악화": [
        "s5_daily_failure_speed", "late_stage_degradation", 
        "error_growth_ratio", "s197_7d_straight_rise"
    ],
    "복구 실패": [
        "recovery_failure_flag"
    ],
    "최근 발생 시점": [
        "timeout_total_days_since_last"
    ],
    "최초 발생 시점": [
        "s187_days_since_first", "s5_days_since_first"
    ],
    "과거 누적 발생": [
        "s5_ever_flag", "s187_ever_flag"
    ],
    "직접 손상 발생": [
        "timeout_5s_damaged", "seek_damaged", "s5_damaged", 
        "s198_damaged", "s187_damaged", "smart_5_raw", "s5_diff"
    ]
}

# ---------------------------------------------------------
# 그룹별 Intra-group Spearman 분석
# ---------------------------------------------------------
threshold = 0.90 

for group_name, cols in feature_groups.items():
    # 1. 컬럼 존재 여부 확인 및 인식되지 않은 컬럼 출력
    valid_cols = [c for c in cols if c in df.columns]
    missing_cols = [c for c in cols if c not in df.columns] # 존재하지 않는 컬럼 리스트업
    
    if missing_cols:
        print(f"⚠️ [{group_name}] 데이터셋에서 찾을 수 없는 컬럼: {missing_cols}")
        
    if len(valid_cols) < 2: 
        continue
    
    # 2. 수치형 변환 및 상수 컬럼 제거
    sub_df = df[valid_cols].apply(pd.to_numeric, errors='coerce')
    sub_df = sub_df.loc[:, sub_df.nunique() > 1] 
    
    # [추가] 값이 하나뿐이라(상수) 제거된 컬럼 확인
    dropped_const = [c for c in valid_cols if c not in sub_df.columns]
    if dropped_const:
        print(f"ℹ️ [{group_name}] 상수(고유값 1개)로 판명되어 분석에서 제외된 컬럼: {dropped_const}")
    
    # 3. Median Imputation (데이터 손실 방지)
    sub_df = sub_df.fillna(sub_df.median())
    if sub_df.empty: continue

    # 4. Spearman 상관계수 계산
    corr_matrix = sub_df.corr(method='spearman')

    print(f"\n" + "="*50)
    print(f"📊 {group_name} (분석 대상: {len(sub_df.columns)}개)")

    # 5. 개별 그룹 시각화 (Heatmap)
    n = len(sub_df.columns)
    figsize = (max(7, min(18, n * 0.7)), max(5, min(14, n * 0.5)))
    plt.figure(figsize=figsize)
    sns.heatmap(corr_matrix, annot=(n <= 15), fmt=".2f", cmap='RdBu_r', vmin=-1, vmax=1, center=0)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # 6. 고중복 페어 탐지
    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    redundant_series = (
        upper.stack()
        .loc[lambda x: abs(x) > threshold]
        .sort_values(key=lambda x: abs(x), ascending=False)
    )

    if not redundant_series.empty:
        print(f"\n[!] 고중복 탐지: {group_name} (|rho| > {threshold})")
        for (feat_a, feat_b), val in redundant_series.items():
            print(f"  - {feat_a} <-> {feat_b} : {val:.4f}")
        print("-" * 50)

```
